```python
import os
from IPython.display import display
import ipywidgets as widgets

# Input widgets
folder_input = widgets.Text(
    value='',
    placeholder='Enter folder path (or file path)',
    description='📂 Folder:',
    layout=widgets.Layout(width='600px')
)

prefix_input = widgets.Text(
    value='',
    placeholder='Enter file prefix',
    description='📝 Prefix:',
    layout=widgets.Layout(width='400px')
)

ext_input = widgets.Text(
    value='',
    placeholder='e.g. .txt, .jpg, .pdf',
    description='🔍 Ext:',
    layout=widgets.Layout(width='300px')
)

run_button = widgets.Button(
    description='🚀 Rename Files',
    button_style='success'
)

output = widgets.Output()

def rename_files(folder_path, prefix, file_extension):
    with output:
        output.clear_output()
        if os.path.isfile(folder_path):
            files_to_rename = [folder_path]
        elif os.path.isdir(folder_path):
            files_to_rename = [
                os.path.join(folder_path, f)
                for f in os.listdir(folder_path)
                if f.endswith(file_extension)
            ]
        else:
            print("❌ Error: Invalid path!")
            return

        if not files_to_rename:
            print("⚠️ No matching files found.")
            return

        for count, file_path in enumerate(files_to_rename, start=1):
            if os.path.isfile(file_path):
                directory, filename = os.path.split(file_path)
                new_name = f"{prefix}{count}{file_extension}"
                new_path = os.path.join(directory, new_name)

                print(f"🔧 Renaming: {file_path} → {new_path}")
                try:
                    os.rename(file_path, new_path)
                    print(f"✅ Renamed: {filename} → {new_name}")
                except Exception as e:
                    print(f"❌ Error renaming {filename}: {e}")
        print("🎉 Bulk renaming complete!")

def on_run_clicked(b):
    folder_path = folder_input.value.strip()
    prefix = prefix_input.value.strip()
    extension = ext_input.value.strip()
    rename_files(folder_path, prefix, extension)

run_button.on_click(on_run_clicked)

# Display widgets
display(folder_input, prefix_input, ext_input, run_button, output)
```
